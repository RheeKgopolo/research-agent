# Research Agent walkthrough
Runs each stage separately so you can inspect outputs. Set `OFFLINE = False` once your keys are in `.env`.

```python
OFFLINE = True
from agent import LLM, plan, research, write, reflect, DEFAULT_MODEL
llm = LLM(DEFAULT_MODEL, offline=OFFLINE)
topic = 'How agentic AI workflows differ from single prompts'
```

```python
steps = plan(llm, topic); print(steps)
```

```python
notes = research(llm, topic, steps); print(notes)
```

```python
draft = write(llm, topic, notes); print(draft)
```

```python
feedback = reflect(llm, topic, draft); print(feedback)
```

```python
from IPython.display import Markdown
final = write(llm, topic, notes, feedback, draft)
Markdown(final)
```
